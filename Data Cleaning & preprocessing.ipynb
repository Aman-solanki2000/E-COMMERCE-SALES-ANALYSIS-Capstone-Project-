#OUTLIERS

```python
def check_outliers(df,column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(f"\n{column:}")
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outliers:", len(outliers))


check_outliers(order_iteams, "quantity")
check_outliers(order_iteams, "unit_price")
check_outliers(order_iteams, "discount(%)")
check_outliers(order_iteams, "shipping_cost")



check_outliers(product, "product_weight_g")
check_outliers(product, "cost_price")
check_outliers(product, "selling_price")


print("Weight min:", product["product_weight_g"].min())
print("Weight max:", product["product_weight_g"].max())

print("Cost Price min:", product["cost_price"].min())
print("Cost Price max:", product["cost_price"].max())

print("Selling Price min:", product["selling_price"].min())
print("Selling Price max:", product["selling_price"].max())
```

```python
product["profit"] = (
    product["selling_price"] - product["cost_price"]
)

product["profit_margin_%"] = (
    product["profit"] / product["selling_price"]
)* 100 

print(product["profit"].describe())
print(product["profit_margin_%"].describe())

print("Negative Profit:",
      (product["profit"] < 0).sum())

print("Zero Profit:",
      (product["profit"] == 0).sum())
```

```python
import pandas as pd
import os

cleaned_folder = "Cleaned Data"

files = [
    "Customers_Cleaned.xlsx",
    "Orders_Cleaned.xlsx",
    "Order_Items_Cleaned.xlsx",
    "Products_Cleaned.xlsx",
    "Reviews_Cleaned.xlsx",
    "Geolocation_Cleaned.xlsx"
]

for file in files:
    excel_path = os.path.join(cleaned_folder, file)
    
    df = pd.read_excel(excel_path)
    
    csv_file = file.replace(".xlsx", ".csv")
    csv_path = os.path.join(cleaned_folder, csv_file)
    
    df.to_csv(csv_path, index=False)
    
    print(f"Created: {csv_file}")

print("\nAll CSV files created successfully!")
```
